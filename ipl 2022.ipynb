#### *Top Scorers :-*

We will identify the players who recorded the highest individual scores in the matches.

 ##### *Insight:-*

This analysis highlights the highest-scoring individual performances recorded in IPL 2022.

#### *10 Best Bowling Figures*

We will identify the 10 best bowling performances using wickets taken and runs conceded.

```python
bowling = df[['best_bowling', 'best_bowling_figure']].copy()

bowling[['wickets', 'runs']] = bowling['best_bowling_figure'].str.extract(
    r'(\d+)\D+(\d+)'
)

bowling['wickets'] = bowling['wickets'].astype(int)
bowling['runs'] = bowling['runs'].astype(int)

bowling.sort_values(
    by=['wickets', 'runs'],
    ascending=[False, True]
).head(10)
```

##### *Explanation:-*
best_bowling contains the bowler's name.
best_bowling_figure contains the bowling figure.
We extract wickets and runs conceded separately.
First, performances are sorted by more wickets.
When wickets are equal, the performance with fewer runs conceded comes first.
head(10) displays the 10 best bowling figures.
##### *Insight:-*

This analysis identifies the bowlers who produced the strongest individual bowling performances in IPL 2022.

## *Visualization Section*

***Visualization 1: Matches Won by Each Team***

A bar chart helps us compare the number of matches won by each IPL team.

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y='match_winner',
    order=df['match_winner'].value_counts().index
)

plt.title('Matches Won by Each Team in IPL 2022')
plt.xlabel('Number of Matches Won')
plt.ylabel('Team')
plt.show()
```

##### *Explanation:-*

countplot() counts the number of matches won by each team.

y='match_winner' displays the team names on the y-axis.

order= arranges the teams from the highest number of wins to the lowest.

plt.title() adds a title to the chart.

plt.show() displays the visualization.

 ##### *Insight:-*

This chart makes it easy to compare team performance and identify the teams with the highest and lowest numbers of wins in IPL 2022.

***Visualization 2: Toss Decision Trends***

We will visualize how often teams chose to Bat or Field after winning the toss.

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='toss_decision',
    order=df['toss_decision'].value_counts().index
)

plt.title('Toss Decision Trends in IPL 2022')
plt.xlabel('Toss Decision')
plt.ylabel('Number of Matches')
plt.show()
```

##### *Explanation:-*
countplot() counts how many times each toss decision was made.
x='toss_decision' displays Bat and Field on the x-axis.
The y-axis shows the number of matches.
 ##### *Insight:-*

The visualization clearly shows that Field was selected more often than Bat after winning the toss in IPL 2022.

***Visualization 3: Toss Winner vs Match Winner***

We will visualize how often the toss winner also won the match.

```python
toss_match = (df['toss_winner'] == df['match_winner']).value_counts()

toss_match.index = ['Toss Winner Lost Match', 'Toss Winner Won Match']

plt.figure(figsize=(8, 5))

sns.barplot(
    x=toss_match.index,
    y=toss_match.values
)

plt.title('Toss Winner vs Match Winner in IPL 2022')
plt.xlabel('Result')
plt.ylabel('Number of Matches')
plt.xticks(rotation=10)
plt.show()
```

##### *Explanation:-*
We compare toss_winner with match_winner.
True means the toss winner also won the match.
False means the toss winner lost the match.
barplot() displays this comparison visually.
##### *Insight:-*

The chart helps us understand the relationship between winning the toss and winning the match in IPL 2022.

***Visualization 4: How Teams Win — Runs vs Wickets***

We will visualize whether teams won their matches by Runs or Wickets.

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='won_by',
    order=df['won_by'].value_counts().index
)

plt.title('How Teams Won Matches in IPL 2022')
plt.xlabel('Won By')
plt.ylabel('Number of Matches')
plt.show()
```

##### *Explanation:-*
countplot() counts the number of matches in each winning category.
won_by contains either Runs or Wickets.
The y-axis represents the number of matches.
#### *Insight:-*

The visualization compares run victories and wicket victories, helping us understand the two main ways teams won matches during IPL 2022.

***Visualization 5: Most Matches Played by Venue***

We will visualize the number of IPL 2022 matches played at each venue.
